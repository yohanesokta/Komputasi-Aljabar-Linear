# Contoh Soal 1 – Harga Tiket

Di sebuah acara kampus, tiket mahasiswa dan tiket umum dijual dengan harga berbeda.

Total 30 tiket terjual dan menghasilkan uang sebesar Rp240.000.
Harga tiket umum Rp4.000 lebih mahal daripada tiket mahasiswa.

Berapakah harga masing-masing tiket?

---

## Variabel

Diketahui:

* ( x ) = harga tiket mahasiswa
* ( y ) = harga tiket umum

---

## Membuat Persamaan

Dari soal:

Jumlah tiket = 30

[
x + y = 30
]

Harga tiket umum lebih mahal Rp4.000

[
y = x + 4000
]

---

## Substitusi

Substitusi persamaan kedua ke pertama:

[
x + (x + 4000) = 240000
]

[
2x + 4000 = 240000
]

[
2x = 236000
]

[
x = 118000
]

Lalu:

[
y = 118000 + 4000
]

[
y = 122000
]

---

## Jawaban

Harga tiket mahasiswa = **Rp118.000**
Harga tiket umum = **Rp122.000**
