# Sistem Persamaan Linear Dua Variabel (SPLDV)

## Contoh Soal 1 – Umur

Umur Andi 4 tahun lebih tua dari umur Budi.  
Jumlah umur mereka sekarang adalah 32 tahun.  
Berapa umur masing-masing?

---

### Diketahui
- x = umur Andi  
- y = umur Budi  

---

$$
\begin{cases}
x + y = 32 \\
x - y = 4
\end{cases}
$$

---

### Penyelesaian

Dari persamaan kedua:

$$
x = y + 4
$$

Substitusi ke persamaan pertama:

$$
(y + 4) + y = 32
$$

$$
2y + 4 = 32
$$

$$
2y = 28
$$

$$
y = 14
$$

$$
x = 18
$$

---

### Jawaban

$$
x = 18
$$

$$
y = 14
$$




## Contoh Soal 2 – Harga Barang

### Soal
Harga 2 buku dan 3 pulpen adalah Rp19.000.  
Harga 1 buku dan 2 pulpen adalah Rp11.000.  
Berapa harga masing-masing?

---

### Misalkan
- x = harga buku  
- y = harga pulpen  

---

### Model Matematika

$$
\begin{cases}
2x + 3y = 19000 \\
x + 2y = 11000
\end{cases}
$$

---

### Penyelesaian

Kalikan persamaan kedua dengan 2:

$$
2x + 4y = 22000
$$

Kurangkan dengan persamaan pertama:

$$
(2x + 4y) - (2x + 3y) = 22000 - 19000
$$

$$
y = 3000
$$

Substitusi ke persamaan kedua:

$$
x + 2(3000) = 11000
$$

$$
x + 6000 = 11000
$$

$$
x = 5000
$$

---

### Jawaban Akhir

$$
x = 5000
$$

$$
y = 3000
$$
